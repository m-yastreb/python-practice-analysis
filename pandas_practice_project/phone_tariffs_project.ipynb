Вопрос 10: Сколько заработала фирма на СМС сообщениях?

Тарифная сетка:
* ВИП - 1 рубль за СМС
* Все за 300 - 2 рубля за СМС
* Всё за 500 - 3 рубля за СМС
* Всё за 800 - 4 рубля за СМС
* Промо - 7 рублей за СМС
* Корпоративный - 0 рублей за СМС
* Всё за 250 (архив) - 5 рублей за СМС

```python
prices = {'ВИП': 1, 'Все за 300': 2, 'Всё за 500': 3, 'Всё за 800': 4, 'Промо': 7, 'Корпоративный': 0,
         'Всё за 250 (архив)': 5}
result = (df.groupby('tarif')['sms'].sum() * pd.Series(prices)).sum()
result
```

Вопрос 17: Поступила задача разбить клиентов на группы
* до 25 лет (включительно) - A (англ. буква)
* с 26 до 40 лет (включительно) - B (англ. буква)
* старше 40 лет - C (англ. буква)

Вопрос 18: Пришла информация, что на сервере в Москве возник сбой и баланс за отправленные смс не списался (у пользователей из Москвы). Спишите с каждого пользователя из Москвы деньги за смс. Количество отправленных смс за которые не списали деньги в столбце sms. Стоимость 1 смс - 3 рубля. (обновите колонку balance)

```python
df['balance'] = np.where(df['city']== 'Moscow', df['balance']-(df['sms'] * 3),df['balance'])
df
```

Вопрос 19: На совещании было принято решение детальнее проработать вопрос списания денежных средств. Теперь, спишите деньги (у пользователей из Москвы за смс) с учетом тарифной сетки:

* ВИП - 1 рубль за СМС
* Все за 300 - 2 рубля за СМС
* Всё за 500 - 3 рубля за СМС
* Всё за 800 - 4 рубля за СМС
* Промо - 7 рублей за СМС
* Корпоративный - 0 рублей за СМС
* Всё за 250 (архив) - 5 рублей за СМС

```python
prices = {'ВИП': 1, 'Все за 300': 2, 'Всё за 500': 3, 'Всё за 800': 4, 'Промо': 7, 'Корпоративный': 0,
         'Всё за 250 (архив)': 5}
df['balance'] = np.where(df['city']== 'Moscow', df['balance'] - (df['sms'] * df['tarif'].map(prices)),df['balance'])
df
```

Вопрос 24: Расcчитайте долю смс для каждого пользователя. Для этого определите сколько всего было отправлено СМС, а затем расcчитайте отношение отправленных смс клиента к общему количеству. Столбец назовите sms_volume. Оставьте TOP-3
